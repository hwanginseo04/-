```python
import numpy as np

# 1. 1,000개 시점의 데이터 시뮬레이션 (A는 변동성이 작고, B는 큼)
np.random.seed(42)
n_periods = 1000
# A: 평균 100, 변동성(std) 5
# B: 평균 100, 변동성(std) 15 (변동성이 훨씬 큼)
A_data = np.random.normal(100, 5, n_periods)
B_data = np.random.normal(100, 15, n_periods)

# 2. 분산(Variance) 계산
var_A = np.var(A_data)
var_B = np.var(B_data)

# 3. 위험 최소화(최소 분산 포트폴리오) 비율 계산
# 상관계수가 0이라고 가정할 때, 분산에 반비례하여 비중을 나눕니다.
# w_A = var_B / (var_A + var_B)
weight_A = var_B / (var_A + var_B)
weight_B = 1 - weight_A

print(f"--- 1,000개 데이터 기준 최적화 결과 ---")
print(f"공급업체 A 분산: {var_A:.2f}")
print(f"공급업체 B 분산: {var_B:.2f}")
print("-" * 30)
print(f"✅ 최적 구매 비율 (공급업체 A): {weight_A*100:.2f}%")
print(f"✅ 최적 구매 비율 (공급업체 B): {weight_B*100:.2f}%")
print("-" * 30)
print("결론: 리스크(분산)를 최소화하려면 변동성이 큰 B의 비중을 낮추고 A를 압도적으로 많이 사야 합니다.")
```

"위험(Risk)의 정의": 분산(Variance)이 곧 리스크입니다. 변동성이 크면 비용 예측이 불가능하므로 리스크가 높다고 판단했습니다.

"최적화 로직": 각 공급업체의 분산에 반비례하도록 가중치를 할당하는 '최소 분산 전략(Minimum Variance Strategy)'을 사용했습니다.

"결과 해석": 변동성이 큰 B업체의 데이터를 1,000개 분석해보니, A보다 위험도가 약 9배(분산값 차이) 높게 나타났습니다. 따라서 위험 분산 차원에서 A에 90%의 비중을 두어 전체 공급 비용의 안정성을 90% 이상 확보하는 전략을 세웠습니다.
