```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.preprocessing import LabelEncoder
import gradio as gr

# 1. 데이터 로드
file_path = 'AdventureWorks Sales.xlsx'
df_sales = pd.read_excel(file_path, sheet_name='Sales_data')
df_prod = pd.read_excel(file_path, sheet_name='Product_data')
df_cust = pd.read_excel(file_path, sheet_name='Customer_data')

# 2. 데이터 전처리 (구매/비구매 데이터 생성 포함)

# 모든 고객-제품 조합 생성
all_customers = df_cust['CustomerKey'].unique()
all_products = df_prod['ProductKey'].unique()
# Use pd.DataFrame.from_records to create a DataFrame from (customer, product) tuples
all_combinations_df = pd.DataFrame({'CustomerKey': np.repeat(all_customers, len(all_products)),
                                    'ProductKey': np.tile(all_products, len(all_customers))})

# 기존 판매 데이터와 병합 (Order Quantity, Sales Amount, Unit Price는 df_sales에서 옴)
# Product, Category, Subcategory 등은 df_prod에서 옴
df_base = pd.merge(all_combinations_df, df_sales, on=['CustomerKey', 'ProductKey'], how='left', suffixes=('_base', '_sales'))
df_base = pd.merge(df_base, df_prod.drop(columns=['Standard Cost', 'Color', 'Size', 'Weight'], errors='ignore'), on='ProductKey', how='left')
df_base = pd.merge(df_base, df_cust, on='CustomerKey', how='left', suffixes=('_base', '_cust')) # Customer info

# 필요한 컬럼만 선택하고, NaN 값 처리
# 'Order Quantity', 'Sales Amount'가 NaN인 경우 0으로 채움 (비구매 건)
df_base['Order Quantity'] = df_base['Order Quantity'].fillna(0)
df_base['Sales Amount'] = df_base['Sales Amount'].fillna(0)

# 'Unit Price' 처리: 비구매 건의 경우 해당 제품의 평균 판매 단가를 사용.
# 먼저 판매 데이터에서 제품별 평균 단가를 계산
product_mean_unit_price = df_sales.groupby('ProductKey')['Unit Price'].mean().reset_index()
product_mean_unit_price.rename(columns={'Unit Price': 'Mean_Unit_Price'}, inplace=True)

# df_base에 제품별 평균 단가 정보를 병합
df_base = pd.merge(df_base, product_mean_unit_price, on='ProductKey', how='left')

# NaN인 'Unit Price' (sales)를 'Mean_Unit_Price'로 채움
df_base['Unit Price'] = df_base['Unit Price'].fillna(df_base['Mean_Unit_Price'])
# 만약 'Mean_Unit_Price'도 NaN이면 (해당 제품이 한 번도 판매되지 않은 경우) 0으로 채움
df_base['Unit Price'] = df_base['Unit Price'].fillna(0)

# 'Product' 컬럼도 NaN이면 'Unknown' 등으로 채움 (극히 드문 경우)
df_base['Product'] = df_base['Product'].fillna('Unknown')

# 최종적으로 필요한 컬럼만 선택하여 df로 사용
df = df_base[['CustomerKey', 'ProductKey', 'Order Quantity', 'Sales Amount', 'Unit Price', 'Product']].copy()
df = df.drop_duplicates()

# 'Buy' 타겟 변수 생성 (Order Quantity가 0보다 크면 1, 아니면 0)
df['Buy'] = df['Order Quantity'].apply(lambda x: 1 if x > 0 else 0)

# 클래스 불균형 해결을 위한 언더샘플링 (Under-sampling)
df_buy_0 = df[df['Buy'] == 0] # 'Buy'가 0인 데이터
df_buy_1 = df[df['Buy'] == 1] # 'Buy'가 1인 데이터

# 'Buy'가 0인 데이터에서 'Buy'가 1인 데이터와 동일한 개수만큼 무작위 샘플링
df_buy_0_sampled = df_buy_0.sample(n=len(df_buy_1), random_state=42)

# 샘플링된 데이터와 'Buy'가 1인 데이터를 합침
df = pd.concat([df_buy_0_sampled, df_buy_1], ignore_index=True)

# 'Buy' 타겟 변수의 분포 확인 (진단을 위해 추가)
print("Distribution of 'Buy' target variable:\n", df['Buy'].value_counts())

# 3. 모델링 준비
le = LabelEncoder()
df['Product'] = df['Product'].astype(str)
le.fit(df['Product']) # Fit on the potentially expanded product list
df['Product_enc'] = le.transform(df['Product'])

# 학습 데이터
X = df[['Unit Price', 'Product_enc']].values
y_clf = df['Buy'].values
y_reg = df['Sales Amount'].values

# 모델 학습
clf = RandomForestClassifier(n_estimators=100, random_state=42)
clf.fit(X, y_clf)
reg = RandomForestRegressor(n_estimators=100, random_state=42)
reg.fit(X, y_reg)

# 4. Gradio 서비스
product_list = sorted(df['Product'].unique().tolist())

def crm_predict(price, product_name):
    try:
        # 인코딩 처리
        if product_name not in le.classes_:
            return "오류: 리스트에 없는 상품입니다."

        p_enc = le.transform([product_name])[0]
        input_val = np.array([[float(price), p_enc]])

        # 예측
        prob = clf.predict_proba(input_val)[0]
        # 클래스가 1개일 경우 대응
        prob_val = prob[1] if len(prob) > 1 else 1.0

        sales = reg.predict(input_val)[0]

        msg = "VIP 타겟팅 추천" if prob_val > 0.5 else "일반 고객 관리"
        return f"구매 확률: {prob_val*100:.1f}% | 예상 매출: ${sales:,.2f} | 전략: {msg}"
    except Exception as e:
        return f"상세 오류: {str(e)}"

# 인터페이스 실행
demo = gr.Interface(
    fn=crm_predict,
    inputs=[
        gr.Number(label="단가(Unit Price)", value=100.0),
        gr.Dropdown(choices=product_list, label="상품명 선택")
    ],
    outputs="text",
    title="AdventureWorks CRM 예측 시스템"
)
demo.launch(share=True)
```
